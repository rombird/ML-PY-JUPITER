```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

# 데이터 준비 (시간에 따른 주가 변화 가정)
# x = np.linspace(0, 10, 100) # 0부터 10까지 100개의 점
# y = np.sin(x) + np.random.normal(0, 0.1, 100)

df = pd.read_csv('CC GENERAL.csv')
x = df['CUST_ID']
y = df['PAYMENTS']

# 그래프 그리기
plt.figure(figsize=(8, 4)) # 캔버스(Figure) 크기 설정
plt.plot(x, y, label='label', color='blue', linestyle='--') # 선 그래프 그리기
plt.title('title') # 제목
plt.xlabel('x_axis') # X축 레이블
plt.ylabel('y_axis') # Y축 레이블
plt.legend('bum') # 범례 표시
plt.grid(True) # 그리드 표시
plt.show() # 그래프 출력
```

```python
import numpy as np
import matplotlib.pyplot as plt

# 데이터 준비 (사람들의 나이 분포 가정)
data_age = np.random.normal(loc=35, scale=10, size=500) # 평균 35, 표준편차 10

plt.figure(figsize=(8, 4))
# 히스토그램 그리기
plt.hist(data_age, bins=10, edgecolor='black', color='skyblue', alpha=0.7) 
plt.title('title')
plt.xlabel('age')
plt.ylabel('frequent')
plt.axvline(np.mean(data_age), color='red', linestyle='dashed', linewidth=1, label=f'mean: {np.mean(data_age):.2f}') # 평균선 추가
plt.legend()
plt.show()
```

```python
# 데이터 준비 (광고비 지출과 매출액 가정)
ad_spend = np.random.uniform(100, 1000, 50)
sales = ad_spend * 2.5 + np.random.normal(0, 500, 50) # 양의 상관관계

plt.figure(figsize=(8, 4))
# 산점도 그리기
plt.scatter(ad_spend, sales, color='purple', alpha=0.6, s=50) # s는 점의 크기
plt.title('Relationship between X and Y')
plt.xlabel('advertising spending(X)')
plt.ylabel('sales(Y)')
plt.show()
```

```python
# Figure(캔버스)와 Axes(각 그래프 영역) 생성
fig, axes = plt.subplots(nrows=1, ncols=2, figsize=(12, 5))

# --- 첫 번째 서브플롯 (히스토그램) ---
axes[0].hist(data_age, bins=15, color='salmon', edgecolor='black')
axes[0].set_title('age distribution')
axes[0].set_xlabel('age')
axes[0].set_ylabel('frequent')

# --- 두 번째 서브플롯 (산점도) ---
axes[1].scatter(ad_spend, sales, color='green', alpha=0.7)
axes[1].set_title('ad vs sales')
axes[1].set_xlabel('ad spending')
axes[1].set_ylabel('sales')

plt.tight_layout() # 서브플롯 간의 간격 자동 조절
plt.show()
```
